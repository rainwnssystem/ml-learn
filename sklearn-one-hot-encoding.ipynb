```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
import numpy as np
```

#### Label Encoding
범주형 데이터에서 각 항목에 고유한 정수를 부여하는 방식이다.

- 학습 시 ML 알고리즘에서 숫자의 대소 관계가 의미 있는 것처럼 해석될 수 있다.
- 예를 들어, Apple=0, Pear=3라면, 알고리즘은 Pear가 더 크다/중요하다 라는 의미로 학습될 수 있고, 이는 model의 예측 성능을 저하시킨다.
- One-hot Encoding이 이러한 Label Encoding의 한계를 보완한다.

```python
items = ['Apple', 'Orange', 'Banana', 'Pear']

encoder = LabelEncoder()
encoder.fit(items)       # fit: 변환 기준 학습
labels = encoder.transform(items)  # transform: 실제 변환
print(labels)
```

#### One-hot Encoding
범주형 데이터에서 각 항목에 대한 새로운 column들을 모두 추가하고, 해당 column의 값을 1로, 나머지 column의 값들은 모두 0으로 표시하여 vector로 표현하는 방식이다.

```python
items = ['Apple', 'Orange', 'Banana', 'Pear']

np.reshape(items, (-1, 1))  # 2차원 array로 reshape
items = np.reshape(items, (-1, 1))
```

```python
encoder = OneHotEncoder()
encoder.fit(items)

labels = encoder.transform(items)  # OneHotEncoder 내부적으로 feature 고유값 확인을 자동으로 수행한다.
labels.toarray()  # OneHotEncoder가 기본적으로 output dtype을 float64로 설정하여, 1. , 0. 과 같이 .(dot)을 포함하여 출력된다.
```

pandas get_dummies()를 사용해서도 One-hot Encoding을 수행할 수 있다.

```python
import pandas as pd

items = pd.DataFrame({'items': ['Apple', 'Orange', 'Banana', 'Pear']})
labels = pd.get_dummies(items)
print(labels)
```
